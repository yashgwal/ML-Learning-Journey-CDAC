```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.linear_model import LinearRegression, ElasticNet,Ridge
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import r2_score
from sklearn.model_selection import KFold,cross_val_score,cross_val_predict, cross_validate,GridSearchCV
from sklearn.pipeline import Pipeline
from sklearn.compose import ColumnTransformer, make_column_selector
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.impute import SimpleImputer
import warnings
from tqdm import tqdm
import os
os.chdir("/home/pgcp-ai/MachineLearning/Datasets/")
```

```python
chem = pd.read_csv('ChemicalProcess.csv')
chem
```

```python
X,y = chem.drop('Yield',axis=1),chem['Yield']
```

```python
imputer = SimpleImputer()
rf = RandomForestRegressor(random_state=26)
kfold = KFold(n_splits=5,shuffle=True,random_state=26)
pipe = Pipeline([('IMP',imputer),('MODEL',rf)])
params = {"MODEL__max_features":[2,5,10,15,25,30,40],
         'IMP__strategy' : ['mean','median']}
gcv = GridSearchCV(estimator=pipe,param_grid=params,cv=kfold,scoring='neg_root_mean_squared_error',n_jobs=-1)
gcv.fit(X,y)
gcv.best_params_,gcv.best_score_
```
