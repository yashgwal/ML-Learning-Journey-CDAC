```python
import pandas as pd
import numpy as np
from sklearn.linear_model import LinearRegression, Lasso, Ridge, ElasticNet
from sklearn.model_selection import train_test_split
from sklearn.metrics import r2_score, mean_absolute_error, mean_squared_error
import os
os.chdir('/home/pgcp-ai/MachineLearning/Datasets')
```

```python
df = pd.read_csv('Concrete_Data.csv')
df
```

```python
X = df.drop('Strength', axis = 1)
y = df['Strength']
```

```python
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=0.3,random_state=26)
```

```python
lr = LinearRegression()
lr.fit(X_train, y_train)
```

```python
y_pred = lr.predict(X_test)
r2_score(y_test, y_pred)
```

```python
alpha_values = np.linspace(0.01, 10)
dict_alpha = dict()
for i in alpha_values:
    lasso = Lasso(alpha = i)
    lasso.fit(X_train, y_train)
    y_pred_lasso = lasso.predict(X_test)
    dict_alpha[i] = r2_score(y_test, y_pred_lasso)
    
lasso_df = pd.DataFrame(dict_alpha.items(), columns = ['Alpha', 'Score'])
lasso_df.sort_values('Score', ascending = False).iloc[0]
```

```python
alpha_values = np.linspace(0.01, 10)
dict_alpha = dict()
for i in alpha_values:
    ridge = Ridge(alpha = i)
    ridge.fit(X_train, y_train)
    y_pred_ridge = ridge.predict(X_test)
    dict_alpha[i] = r2_score(y_test, y_pred_ridge)
    
ridge_df = pd.DataFrame(dict_alpha.items(), columns = ['Alpha', 'Score'])
ridge_df.sort_values('Score', ascending = False).iloc[0]
```

```python
alpha_values = np.linspace(0.01, 10)
l1_ratios = np.linspace(0,1)
dict_alpha = dict()
scores = []
for i,j in zip(alpha_values,l1_ratios):
    el = ElasticNet(alpha = i,l1_ratio=j)
    el.fit(X_train, y_train)
    y_pred_ElasticNet = el.predict(X_test)
    dict_alpha[i] = r2_score(y_test, y_pred_ElasticNet)
    scores.append([i,j, r2_score(y_test, y_pred_ElasticNet)])
# ElasticNet_df = pd.DataFrame(dict_alpha.items(), columns = ['Alpha', 'Score'])
# ElasticNet_df.sort_values('Score', ascending = False)

df1 = pd.DataFrame(scores, columns = ['Alpha', 'L1 Ratio', 'Score'])
df1.sort_values('Score', ascending = False).iloc[0]
```
