```python
import pandas as pd
import numpy as np
from statsmodels.graphics.tsaplots import plot_acf
import matplotlib.pyplot as plt
import statsmodels.api as sm
from statsmodels.tsa.stattools import adfuller
import os
os.chdir("C:/Users/PGCP-AI/ML/MachineLearning/Datasets")
```

```python
df = pd.read_csv("sales-of-shampoo-over-a-three-ye.csv")
df
```

```python
plt.figure(figsize=(8,6))
df.plot()
plt.title('DF Before Differencing')
plt.grid(True)
plt.xlabel('Time Interval')
plt.ylabel('Shampoo Sales')
plt.show()
```

### Before Differencing

```python
result = adfuller(df['Sales of shampoo over a three year period'], maxlag=10)
print("P-Value =", result[1])
if result[1] < 0.05:
    print("Time Series is Stationary")
else:
    print("Time Series is not Stationary")
```

### After Differencing

```python
differenced = df['Sales of shampoo over a three year period'].diff()
diff = differenced[1:]
result = adfuller(diff, maxlag=10)
print("P-Value =", result[1])
if result[1] < 0.05:
    print("Time Series is Stationary")
else:
    print("Time Series is not Stationary")
```

```python
plt.figure(figsize=(8,6))
diff.plot()
plt.title('DF After Differencing')
plt.grid(True)
plt.xlabel('Time Interval')
plt.ylabel('Shampoo Sales')
plt.show()
```
