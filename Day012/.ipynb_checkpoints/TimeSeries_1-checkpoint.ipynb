```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import os
os.chdir("C:/Users/PGCP-AI/ML/MachineLearning/Datasets")
```

```python
df = pd.read_csv("monthly-milk-production-pounds-p.csv", index_col = 0)
df.index = pd.to_datetime(df.index).to_period('M')
df
```

```python
df.plot()
plt.show()
```

```python
downsampled = df.resample('Q').sum()
downsampled.index.rename('Quarter', inplace = True)
downsampled.plot()
plt.title("Quarterly Milk Production")
plt.xlabel("Quarter")
plt.show()
```

```python
from statsmodels.tsa.seasonal import seasonal_decompose
```

```python
result = seasonal_decompose(df['Milk'].values, model = 'additive', period = 12)
result.plot()
plt.show()
```

```python
result = seasonal_decompose(df['Milk'].values, model = 'multiplicative', period = 4)
result.plot()
plt.show()
```

```python
result = seasonal_decompose(downsampled['Milk'].values, model = 'additive', period = 28)
result.plot()
plt.show()
```

```python
from ipywidgets import interact, widgets
```

```python
def CMA(span):
    fcast = df["Milk"].rolling(span, center = True).mean()
    plt.plot(df["Milk"], label = "Data")
    plt.plot(fcast, label = "Centered Rolling Mean")
    plt.legend(loc = 'best')
    plt.show()
interact(CMA, span=(1, 6))
```
