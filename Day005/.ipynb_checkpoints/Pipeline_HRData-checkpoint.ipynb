```python
hr = pd.read_csv("HR_comma_sep.csv")
hr
```

```python
hr.isna().sum()
```

```python
X, y = hr.drop('left', axis = 1), hr['left']
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size = 0.3, random_state = 26,stratify=hr['left'])
```

```python
ohe = OneHotEncoder(sparse_output = False, drop = 'first',handle_unknown='ignore').set_output(transform = 'pandas')
transformer = ColumnTransformer(transformers=[('OHE',ohe,make_column_selector(dtype_include=object)),
                                            ]
                               ,remainder='passthrough',
                               verbose_feature_names_out=False
                              ).set_output(transform='pandas')
```

```python
X_train_trans = transformer.fit_transform(X_train)
X_test_trans = transformer.transform(X_test)
```

```python
ss = StandardScaler()
```

```python
ohe = OneHotEncoder(sparse_output = False, drop = 'first',handle_unknown='ignore').set_output(transform = 'pandas')
transformer = ColumnTransformer(transformers=[('OHE',ohe,make_column_selector(dtype_include=object)),
                                            ]
                               ,remainder='passthrough',
                               verbose_feature_names_out=False
                              ).set_output(transform='pandas')

ss = StandardScaler()
svm = SVC(kernel = 'linear', probability = True, random_state = 26, C = 3.947579)
pipe = Pipeline([('OHE', ohe), ('SS', ss), ('SVM', svm)])
pipe.fit(X_train, y_train)
y_pred = pipe.predict(X_test)
accuracy_score(y_test, y_pred)
```
