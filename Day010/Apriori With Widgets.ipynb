```python
import pandas as pd
from mlxtend.frequent_patterns import apriori, association_rules
import os
import warnings 
warnings.filterwarnings('ignore')
os.chdir("/home/pgcp-ai/MachineLearning/Datasets/")
```

```python
fp_df = pd.read_csv("Faceplate.csv", index_col = 0)
fp_df = fp_df.astype(bool)
itemsets = apriori(fp_df, min_support = 0.2, use_colnames = True)
rules = association_rules(itemsets, metric = 'confidence', min_threshold = 0.6)
rules = rules[['antecedents', 'consequents', 'support', 'confidence', 'lift']]
relv_rules = rules[rules['lift'] > 1]
relv_rules.sort_values('lift', ascending = False)
```

```python
from ipywidgets import interact,widgets
```

```python
cosmetics = pd.read_csv('Cosmetics.csv',index_col=0)
cosmetics
```

```python
def cosmetic_fun(Min_Support,Min_Threshold):
    itemsets = apriori(cosmetics, min_support = Min_Support, use_colnames = True)
    rules = association_rules(itemsets, metric = 'confidence', min_threshold = Min_Threshold)
    rules = rules[['antecedents', 'consequents', 'support', 'confidence', 'lift']]
    relv_rules = rules[rules['lift'] > 1]
    if len(relv_rules) == 0:
        return f"No Relevant Rules Found!!"
    else:
        return relv_rules.sort_values('lift',ascending=False)
```

```python
interact(cosmetic_fun, Min_Support = widgets.FloatSlider(min = 0.01, max = 1, step = 0.01, value = 0.3,description="Minimum Support: "),
         Min_Threshold = widgets.FloatSlider(min = 0.01, max = 1, step = 0.01, value = 0.3,description="Minimum Threshold: "))
```
