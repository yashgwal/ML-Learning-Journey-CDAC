```python
chem = pd.read_csv("ChemicalProcess.csv")
chem
```

```python
chem.isna().sum().sum()
```

```python
chem.mean()
```

```python
chem_mean_imp = SimpleImputer(strategy = 'mean').set_output(transform='pandas')
chem_mean_imp.fit(chem)
```

```python
chem = chem_mean_imp.transform(chem)
```

```python
chem
X = chem.drop('Yield',axis=1)
y = chem['Yield']
```

```python
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=0.3,random_state=26)
```

```python
lr = LinearRegression()
lr.fit(X_train,y_train)
```

```python
y_pred = lr.predict(X_test)
```

```python
print(f'R2 Score : {r2_score(y_test,y_pred)}')
```

```python
y_pred = lr.predict(X_test)
```

### The following is the way the above code should be written

```python
chem = pd.read_csv("ChemicalProcess.csv")
```

```python
X = chem.drop('Yield', axis = 1)
y = chem['Yield']
X.isnull().sum(),y.isnull().sum()
```

```python
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=0.3,random_state=26)
mean_imp = SimpleImputer(strategy = 'mean').set_output(transform='pandas')
X_train_imp = mean_imp.fit_transform(X_train)
X_test_imp = mean_imp.transform(X_test)
```

```python
lr = LinearRegression()
lr.fit(X_train_imp,y_train)
```

```python
y_pred = lr.predict(X_test_imp)
```

```python
print(f'R2 Score : {r2_score(y_test,y_pred)}')
```

### Median

```python
chem = pd.read_csv("ChemicalProcess.csv")
chem
```

```python
X = chem.drop('Yield', axis = 1)
y = chem["Yield"]
```

```python
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=0.3,random_state=26)
median_imp = SimpleImputer(strategy = 'median').set_output(transform='pandas')
X_train_imp = median_imp.fit_transform(X_train)
X_test_imp = median_imp.transform(X_test)
```

```python
lr = LinearRegression()
lr.fit(X_train_imp, y_train)
```

```python
y_pred = lr.predict(X_test_imp)
```

```python
print(f"R2 Score = {r2_score(y_test, y_pred)}")
```
