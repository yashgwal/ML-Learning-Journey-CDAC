```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import LabelEncoder, OneHotEncoder, StandardScaler
from sklearn.model_selection import train_test_split
from sklearn.metrics import log_loss, accuracy_score, f1_score, confusion_matrix, precision_score, recall_score, classification_report,roc_auc_score,roc_curve,RocCurveDisplay
import os
from sklearn.impute import SimpleImputer
import matplotlib.pyplot as plt
from sklearn.svm import SVC
from tqdm import tqdm
from sklearn.pipeline import Pipeline
os.chdir('/home/pgcp-ai/MachineLearning/Cases/Image_Segmentation/')
```

#### Without Pipeline

```python
image = pd.read_csv("Image_Segmentation.csv")
image
```

```python
le = LabelEncoder()
```

```python
image['Class'] = le.fit_transform(image['Class'])
```

```python
X,y = image.drop('Class',axis=1),image['Class']
```

```python
scaler = StandardScaler()
svm = SVC(C=0.264105,kernel='linear',decision_function_shape='ovr')
pipe = Pipeline([('SCL',scaler),('SVM',svm)])
pipe.fit(X,y)
```

```python
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=0.3,stratify=y,random_state=26)
```

```python
X_train_sc = scaler.fit_transform(X_train)
X_test_sc = scaler.transform(X_test)
svm = SVC(C=0.264105,kernel='linear',decision_function_shape='ovr',random_state=26)
svm.fit(X_train_sc, y_train)
y_pred = svm.predict(X_test_sc)
accuracy_score(y_test, y_pred)
```

```python
pipe.fit(X_train,y_train)
y_pred = pipe.predict(X_test)
accuracy_score(y_test,y_pred)
```
