```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from tqdm import tqdm
from sklearn.preprocessing import LabelEncoder,OneHotEncoder,StandardScaler,MinMaxScaler
from sklearn.linear_model import LogisticRegression
from sklearn.naive_bayes import GaussianNB
from sklearn.neighbors import KNeighborsClassifier
from sklearn.ensemble import RandomForestClassifier,VotingClassifier
from sklearn.svm import SVC
from sklearn.compose import ColumnTransformer,make_column_selector
from sklearn.impute import SimpleImputer
from sklearn.model_selection import train_test_split
from sklearn.pipeline import Pipeline
from sklearn.metrics import log_loss,accuracy_score,f1_score,balanced_accuracy_score
import os
from xgboost import XGBClassifier
from lightgbm import LGBMClassifier
from catboost import CatBoostClassifier
os.chdir('/home/pgcp-ai/MachineLearning/Datasets/IrrigationNeed/')
```

```python
df = pd.read_csv('train.csv',index_col='id')
df
```

```python
df.info()
```

```python
ohe = OneHotEncoder(sparse_output=False,drop='first',handle_unknown='ignore').set_output(transform='pandas')
ss = StandardScaler().set_output(transform='pandas')
```

```python
transformer = ColumnTransformer(transformers = [('OHE',ohe,make_column_selector(dtype_include=object))],
                               remainder='passthrough',
                               verbose_feature_names_out=False,
                               )
```

```python
X,y = df.drop('Irrigation_Need',axis=1), df['Irrigation_Need']
```

```python
le = LabelEncoder()
```

```python
y = le.fit_transform(y)
```

```python
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=0.2,stratify=y)
```

```python
X_train_ohe = transformer.fit_transform(X_train)
X_test_ohe = transformer.transform(X_test)
```

```python
X_train_scaled = ss.fit_transform(X_train_ohe)
X_test_scaled = ss.transform(X_test_ohe)
```

```python
X_train_scaled
```

```python
Cs = np.linspace(0.01,100,20)
scores = []
for c in tqdm(Cs):
    lr = LogisticRegression(C=c,solver='lbfgs')
    lr.fit(X_train_scaled,y_train)
    y_pred_prob = lr.predict_proba(X_test_scaled)
    y_pred = lr.predict(X_test_scaled)
    scores.append(['lbfgs',c,log_loss(y_test,y_pred_prob),accuracy_score(y_test,y_pred),f1_score(y_true=y_test,y_pred=y_pred,average='weighted')])
df_scores = pd.DataFrame(scores,columns=['Solver','C','Log-Loss','Accuracy','F1-Score'])
df_scores.sort_values(['Log-Loss','F1-Score','Accuracy'],ascending=[True,False,False])
```

```python
Cs = np.linspace(0.01,100,20)
scores = []
for c in tqdm(Cs):
    lr = LogisticRegression(C=c,solver='newton-cg')
    lr.fit(X_train_scaled,y_train)
    y_pred_prob = lr.predict_proba(X_test_scaled)
    y_pred = lr.predict(X_test_scaled)
    scores.append(['newton-cg',c,log_loss(y_test,y_pred_prob),accuracy_score(y_test,y_pred),f1_score(y_true=y_test,y_pred=y_pred,average='weighted')])
df_scores = pd.DataFrame(scores,columns=['Solver','C','Log-Loss','Accuracy','F1-Score'])
df_scores.sort_values(['Log-Loss','F1-Score','Accuracy'],ascending=[True,False,False])
```

```python
Cs = np.linspace(0.01,100,20)
scores = []
for c in tqdm(Cs):
    lr = LogisticRegression(C=c,solver='sag')
    lr.fit(X_train_scaled,y_train)
    y_pred_prob = lr.predict_proba(X_test_scaled)
    y_pred = lr.predict(X_test_scaled)
    scores.append(['sag',c,log_loss(y_test,y_pred_prob),accuracy_score(y_test,y_pred),f1_score(y_true=y_test,y_pred=y_pred,average='weighted')])
df_scores = pd.DataFrame(scores,columns=['Solver','C','Log-Loss','Accuracy','F1-Score'])
df_scores.sort_values(['Log-Loss','F1-Score','Accuracy'],ascending=[True,False,False])
```

```python
Cs = np.linspace(0.01,100,20)
scores = []
for c in tqdm(Cs):
    lr = LogisticRegression(C=c,solver='newton-cholesky')
    lr.fit(X_train_scaled,y_train)
    y_pred_prob = lr.predict_proba(X_test_scaled)
    y_pred = lr.predict(X_test_scaled)
    scores.append(['newton-cholesky',c,log_loss(y_test,y_pred_prob),accuracy_score(y_test,y_pred),f1_score(y_true=y_test,y_pred=y_pred,average='weighted')])
df_scores = pd.DataFrame(scores,columns=['Solver','C','Log-Loss','Accuracy','F1-Score'])
df_scores.sort_values(['Log-Loss','F1-Score','Accuracy'],ascending=[True,False,False])
```

```python
noOfNeighbors = np.arange(1,16)
scores = []
for n in tqdm(noOfNeighbors):
    knn = KNeighborsClassifier(n_neighbors=n,n_jobs=-1)
    knn.fit(X_train_scaled,y_train)
    y_pred_prob = knn.predict_proba(X_test_scaled)
    y_pred = knn.predict(X_test_scaled)
    scores.append([n,log_loss(y_test,y_pred_prob),accuracy_score(y_test,y_pred),f1_score(y_true=y_test,y_pred=y_pred,average='weighted')])
df_scores = pd.DataFrame(scores,columns=['No of Neighbors','Log-Loss','Accuracy','F1-Score'])
df_scores.sort_values(['Log-Loss','F1-Score','Accuracy'],ascending=[True,False,False]) 
```

```python
nb = GaussianNB()
nb.fit(X_train_scaled,y_train)
y_pred = nb.predict(X_test_scaled)
y_pred_prob = nb.predict_proba(X_test_scaled)
print(f"Accuracy: {accuracy_score(y_test,y_pred)}")
print(f"F1-Score: {f1_score(y_true=y_test,y_pred=y_pred,average='weighted')}")
print(f"Log-Loss: {log_loss(y_test,y_pred_prob)}")
```
