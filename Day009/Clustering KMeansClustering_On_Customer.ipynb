```python
rfm = pd.read_csv('rfm_data_customer.csv', index_col = 0)
rfm
```

```python
X = rfm[["recency_days", "number_of_orders", "revenue"]]
X
```

```python
X.isna().sum()
```

```python
ss = StandardScaler().set_output(transform='pandas')
X_scaled = ss.fit_transform(X)
```

```python
scores = []
no_of_clusters = [2,3,4,5,6]

for n in no_of_clusters:
    clust = KMeans(n_clusters = n, random_state = 26)
    clust.fit(X_scaled)
    score = silhouette_score(X_scaled, clust.labels_)
    scores.append([n, score])
    
df_scores = pd.DataFrame(scores, columns = ["Number of Clusters", "Silhouette Score"])
df_scores.sort_values('Silhouette Score', ascending = False)
```

```python
df = X.copy()
km = KMeans(n_clusters = 3, random_state = 26)
km.fit(X_scaled)
df['Cluster'] = km.labels_
df.groupby('Cluster').mean()
```
