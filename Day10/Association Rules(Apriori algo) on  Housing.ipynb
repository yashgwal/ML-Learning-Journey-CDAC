```python
import pandas as pd
from ipywidgets import widgets, interact
import os
from mlxtend.frequent_patterns import apriori, association_rules
os.chdir('/home/pgcp-ai/MachineLearning/Datasets/')
```

```python
housing = pd.read_csv('Housing.csv')
intervals = [(0,50000),(50000,100000),(100000,150000),(150000,200000)]
bins = pd.IntervalIndex.from_tuples(intervals)
housing['price_slab'] = pd.cut(housing['price'],bins)
intervals = [(0,5000),(5000,10000),(10000,15000),(15000,20000)]
bins = pd.IntervalIndex.from_tuples(intervals)
housing['area_slab'] = pd.cut(housing['lotsize'],bins)
housing.drop(['price','lotsize'],axis=1,inplace=True)
housing = housing.astype(object)
fp_df = pd.get_dummies(housing,prefix_sep='=')
```

```python
def grocery_fun(Min_Support,Min_Threshold):
    itemsets = apriori(fp_df, min_support = Min_Support, use_colnames = True)
    rules = association_rules(itemsets, metric = 'confidence', min_threshold = Min_Threshold)
    rules = rules[['antecedents', 'consequents', 'support', 'confidence', 'lift']]
    relv_rules = rules[rules['lift'] > 1]
    if len(relv_rules) == 0:
        return f"No Relevant Rules Found!!"
    else:
        return relv_rules.sort_values('lift',ascending=False)
```

```python
interact(grocery_fun, Min_Support = widgets.FloatSlider(min = 0.01, max = 1, step = 0.01, value = 0.3,description="Minimum Support: "),
         Min_Threshold = widgets.FloatSlider(min = 0.01, max = 1, step = 0.01, value = 0.3,description="Minimum Threshold: "))
```
