```python
import numpy as np
import pandas as pd
from sklearn.preprocessing import OneHotEncoder
import os
from sklearn.compose import ColumnTransformer
os.chdir('/home/pgcp-ai/MachineLearning/Datasets')
```

```python
tel = pd.read_csv("Telecom.csv")
```

```python
ohe = OneHotEncoder(sparse_output = False, drop='first').set_output(transform='pandas')
ohe.fit_transform(tel)
```

```python
housing = pd.read_csv('Housing.csv')
```

```python
# ohe = OneHotEncoder(sparse_output = False, drop='first').set_output(transform='pandas')
housing
```

```python
housing.info()
list(housing.select_dtypes(include = 'object').columns)
```

```python
transformer = ColumnTransformer(transformers=[("OHE",ohe,['driveway', 'recroom', 'fullbase', 'gashw', 'airco', 'prefarea'])], 
                               remainder = 'passthrough', verbose_feature_names_out = False).set_output(transform='pandas')
transformer.fit_transform(housing)
```

```python
from sklearn.preprocessing import StandardScaler
```

```python
ss = StandardScaler()
```

```python
transformer_1 = ColumnTransformer(transformers=[("OHE",ohe,list(housing.select_dtypes(include='object'))),("SS",ss,['lotsize','price'])
                                               ],
                                 remainder='passthrough',
                                 verbose_feature_names_out=False).set_output(transform='pandas')
housing = transformer_1.fit_transform(housing)
```

```python
from sklearn.compose import make_column_selector
```

```python
transformer2 = ColumnTransformer(transformers=[("OHE",ohe, make_column_selector(dtype_include=object) )], 
                               remainder = 'passthrough', verbose_feature_names_out = False).set_output(transform='pandas')
transformer2.fit_transform(housing)
```
