```python
import pandas as pd
import numpy as np
from sklearn.preprocessing import LabelEncoder,OneHotEncoder, StandardScaler, MinMaxScaler
from sklearn.naive_bayes import GaussianNB, BernoulliNB
from sklearn.model_selection import train_test_split
from sklearn.compose import ColumnTransformer, make_column_selector
from sklearn.metrics import log_loss, accuracy_score, f1_score, confusion_matrix, precision_score, recall_score, classification_report,roc_auc_score,roc_curve,RocCurveDisplay
import os
from sklearn.impute import SimpleImputer
import matplotlib.pyplot as plt
```

```python
irrigation = pd.read_csv("train.csv", index_col = 0)
irrigation
```

```python
irrigation.info()
```

```python
irrigation.isna().sum().sum()
```

```python
le = LabelEncoder()
irrigation["Irrigation_Need"] = le.fit_transform(irrigation["Irrigation_Need"])
irrigation
```

```python
X, y = irrigation.drop("Irrigation_Need", axis = 1), irrigation["Irrigation_Need"]
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size = 0.3, random_state = 26,stratify=irrigation['Irrigation_Need'])
```

```python
ohe = OneHotEncoder(sparse_output = False, drop = 'first').set_output(transform = 'pandas')
transformer = ColumnTransformer(transformers=[('OHE',ohe,make_column_selector(dtype_include=object)),
                                            ]
                               ,remainder='passthrough',
                               verbose_feature_names_out=False
                               ).set_output(transform='pandas')
```

```python
X_train_ohe = transformer.fit_transform(X_train)
X_test_ohe = transformer.transform(X_test)
```

```python
ss = StandardScaler()
X_train_scaled = ss.fit_transform(X_train_ohe)
X_test_scaled = ss.transform(X_test_ohe)
```

```python
bm = GaussianNB()
bm.fit(X_train_scaled, y_train)
y_pred = bm.predict(X_test_scaled)
y_pred_proba = bm.predict_proba(X_test_scaled)

y_pred_labels = le.inverse_transform(y_pred)
y_pred_labels
```

```python
print(log_loss(y_test, y_pred_proba))
```

```python
print(accuracy_score(y_test, y_pred))
```

```python
print(classification_report(y_test, y_pred))
```

```python
tst = pd.read_csv("test.csv", index_col = 0)
tst_ohe = transformer.transform(tst)
tst_scaled = ss.transform(tst_ohe)
tst_pred = bm.predict(tst_scaled)
tst_pred
```

```python
submission = pd.read_csv("sample_submission.csv")
submission
```

```python
submission["Irrigation_Need"] = le.inverse_transform(tst_pred)
submission
```
