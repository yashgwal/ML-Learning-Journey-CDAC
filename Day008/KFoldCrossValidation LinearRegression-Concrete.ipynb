```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.linear_model import LinearRegression, ElasticNet,Ridge
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import r2_score
from sklearn.model_selection import KFold,cross_val_score,cross_val_predict, cross_validate,GridSearchCV
from sklearn.pipeline import Pipeline
from sklearn.compose import ColumnTransformer, make_column_selector
from sklearn.preprocessing import OneHotEncoder, StandardScaler
import warnings
from tqdm import tqdm
import os
os.chdir("/home/pgcp-ai/MachineLearning/Cases/Concrete_Strength/")
```

```python
concrete = pd.read_csv("Concrete_Data.csv")
concrete
```

```python
X, y = concrete.drop("Strength", axis = 1), concrete["Strength"]
```

```python
lr = LinearRegression()
kfold = KFold(n_splits=5,shuffle=True,random_state=26)
results = cross_val_score(lr,X,y,cv=kfold)
print(results)
print(np.mean(results))
```

```python
y_pred = cross_val_predict(lr,X,y)
y_pred
```

```python
cross_validate(lr, X, y)
```

```python
alphas = np.linspace(0.001, 10, 30)
scores = []
folds = KFold(n_splits = 5, shuffle = True, random_state = 26)

for a in tqdm(alphas):
    ridge = Ridge(alpha = a)
    score = cross_val_score(ridge, X, y, cv=folds)
    print(score)
    scores.append([a, np.mean(score)])
    
df_scores = pd.DataFrame(scores, columns = ["Alpha", "Cross Validation Score"])
df_scores.sort_values("Cross Validation Score", ascending = False)
```

```python
alphas = np.linspace(0.001, 10, 30)
ratios = np.linspace(0.001, 1, 10)
scores = []
folds = KFold(n_splits = 5, shuffle = True, random_state = 26)

for a in tqdm(alphas):
    for r in ratios:
        el = ElasticNet(alpha = a, l1_ratio = r)
        score = cross_val_score(el, X, y, cv=folds)
#         print(score)
        scores.append([a, np.mean(score)])
    
df_scores = pd.DataFrame(scores, columns = ["Alpha", "Cross Validation Score"])
df_scores.sort_values("Cross Validation Score", ascending = False)
```

```python
alphas = np.linspace(0.001, 10, 30)
ratios = np.linspace(0.001, 1, 10)
scores = []
folds = KFold(n_splits = 5, shuffle = True, random_state = 26)

for a in tqdm(alphas):
    for r in ratios:
        el = ElasticNet(alpha = a, l1_ratio = r)
        score = cross_val_score(el, X, y, cv=folds,scoring='neg_root_mean_squared_error')
#         print(score)
        scores.append([a, np.mean(score)])
    
df_scores = pd.DataFrame(scores, columns = ["Alpha", "Cross Validation Score"])
df_scores.sort_values("Cross Validation Score", ascending = False)
```

```python
max_features = np.arange(1,len(X.columns)+1)
scores = []
folds = KFold(n_splits = 5, shuffle = True, random_state = 26)

for m in tqdm(max_features):
    
    rf = RandomForestRegressor(max_features=m,random_state=26)
    score = cross_val_score(rf, X, y, cv=folds,scoring='neg_root_mean_squared_error')
#         print(score)
    scores.append([m, np.mean(score)])
    
df_scores = pd.DataFrame(scores, columns = ["Maximum Features", "Cross Validation Score"])
df_scores.sort_values("Cross Validation Score", ascending = False)
```

```python
kfold = KFold(n_splits = 5, shuffle = True, random_state = 26)
max_features = [2,3,4,5,6,7]
params = {'max_features' : max_features}
rf = RandomForestRegressor(random_state=26)
gcv = GridSearchCV(estimator=rf,param_grid=params,cv=kfold,scoring='neg_root_mean_squared_error',n_jobs = -1, verbose=3)
gcv.fit(X,y)
gcv.best_score_,gcv.best_params_
```

```python
housing = pd.read_csv("Housing.csv")
housing
```

```python
X,y = housing.drop('price',axis=1),housing['price']
```

```python
ohe = OneHotEncoder(drop = 'first', handle_unknown = 'ignore')
transformer = ColumnTransformer(transformers = [('OHE', ohe, make_column_selector(dtype_include = object))]
                               , remainder='passthrough', verbose_feature_names_out=False)
```

```python
max_features = [2,3,4,5,6,7]
scores = []
folds = KFold(n_splits = 5, shuffle = True, random_state = 26)

for m in tqdm(max_features):
    rf = RandomForestRegressor(max_features=m,random_state=26)
    pipe = Pipeline([('OHE',transformer),('RF',rf)])
    score = cross_val_score(pipe, X, y, cv=folds,scoring='neg_root_mean_squared_error')
#         print(score)
    scores.append([m,np.mean(score)])
    
df_scores = pd.DataFrame(scores, columns = ["Maximum Features", "Cross Validation Score"])
df_scores.sort_values("Cross Validation Score", ascending = False)
```
