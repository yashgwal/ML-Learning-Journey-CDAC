```python
from sentence_transformers import SentenceTransformer
from tqdm import tqdm
import pandas as pd
import numpy as np
import os
os.chdir('C:/Users/PGCP-AI/ML/MachineLearning/Cases_Rec_Sys/Amazon_Sales/')
```

```python
model = SentenceTransformer("BAAI/bge-m3")
```

```python
df = pd.read_csv('amazon.csv',index_col='product_id')
df
```

```python
about_list = list(df['about_product'])
about_embeddings = []
```

```python
for i in tqdm(about_list):
    embedding = model.encode(i)
    about_embeddings.append(embedding)
about_embeddings = np.array(about_embeddings)
```

```python
print(about_embeddings)
```

```python
sims = []
query = "Good quality gaming setup, good audio and video quality"
query_embedding = model.encode([query])
for e in tqdm(about_embeddings):
    sims.append(model.similarity(query_embedding,e).numpy()[0])
df1 = df[['product_name','about_product','category']].copy()
df1["sims"] = np.array(sims)
df1.sort_values('sims', ascending = False)[:5]
# sims
```
