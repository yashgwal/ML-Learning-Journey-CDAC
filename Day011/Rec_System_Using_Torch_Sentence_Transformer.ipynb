```python
df = pd.read_csv('titles.csv')
df
```

```python
df_indian_movies = df[(df['production_countries'].str.contains('IN')) & (df['type'] == 'MOVIE') & (df['release_year'] >= 1980)]
df_indian_movies.shape
```

```python
all_desc = list(df_indian_movies['description'])
out_embs = []
for r in tqdm(all_desc):
    embd = model.encode(r)
    out_embs.append(embd)

out_embs = np.array(out_embs)
out_embs.shape
```

```python
query = "detective investigating a robbery"
quer1 = ""
qy_emb = model.encode([query])
sims = []

for e in tqdm(out_embs):
    sims.append(model.similarity(qy_emb, e).numpy()[0])
df_ind = df_indian_movies[['id', 'title', 'description', 'release_year']].copy()
df_ind['sims'] = np.array(sims)
df_ind.sort_values('sims', ascending = False)[:5]
```
